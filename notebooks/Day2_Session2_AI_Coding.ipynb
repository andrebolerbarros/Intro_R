### 3.1 Code Generation

Let's see how AI can help generate code for common tasks.

**Example Task**: "Create a function that reads multiple CSV files from a directory 
and combines them into a single data frame, adding a column indicating the source file."

````r
# AI-generated code example (you would get this from AI)
read_and_combine_csvs <- function(directory_path, pattern = "\\.csv$") {
  # Get list of CSV files
  file_list <- list.files(path = directory_path, 
                          pattern = pattern, 
                          full.names = TRUE)
  
  # Check if any files found
  if (length(file_list) == 0) {
    stop("No CSV files found in the specified directory")
  }
  
  # Read and combine files
  combined_data <- do.call(rbind, lapply(file_list, function(file) {
    df <- read.csv(file)
    df$source_file <- basename(file)
    return(df)
  }))
  
  return(combined_data)
}

# Test the function (you would create test files first)
# combined <- read_and_combine_csvs("my_data_folder")
````

### 3.2 Code Explanation

AI can help understand complex code. Let's look at an example.

**Example**: Understanding this dplyr pipeline

````r
# Complex code that might be confusing
library(dplyr)

result <- iris %>%
  group_by(Species) %>%
  summarise(
    n = n(),
    mean_petal_length = mean(Petal.Length),
    sd_petal_length = sd(Petal.Length),
    .groups = "drop"
  ) %>%
  mutate(
    cv = sd_petal_length / mean_petal_length * 100
  ) %>%
  arrange(desc(mean_petal_length))

print(result)
````

**AI Explanation of the above code:**

This code performs a grouped analysis of the iris dataset:

1. `group_by(Species)`: Groups data by iris species
2. `summarise()`: Calculates for each species:
   - `n()`: Count of observations
   - `mean(Petal.Length)`: Average petal length
   - `sd(Petal.Length)`: Standard deviation of petal length
3. `mutate()`: Adds coefficient of variation (CV) as percentage
4. `arrange(desc())`: Sorts by mean petal length in descending order

**Output**: A summary table showing species statistics ordered by petal length.

### 3.3 Debugging with AI

AI can help identify and fix errors in your code.

**Example**: Debugging a function with an error

````r
# Original code with a bug
calculate_percentage <- function(part, total) {
  percentage <- part / total * 100
  return(paste(percentage, "%"))
}

# This works fine
calculate_percentage(25, 100)

# But what if we want to use it with vectors?
parts <- c(10, 20, 30)
totals <- c(100, 100, 100)
# calculate_percentage(parts, totals)  # Try this - what happens?
````

````r
# AI-suggested fix: Vectorize the function properly
calculate_percentage_fixed <- function(part, total) {
  percentage <- (part / total) * 100
  # Use vectorized paste0 for cleaner output
  result <- paste0(round(percentage, 1), "%")
  return(result)
}

# Now it works with vectors
parts <- c(10, 20, 30)
totals <- c(100, 100, 100)
calculate_percentage_fixed(parts, totals)
````

### 3.4 Code Optimization

AI can suggest more efficient ways to write code.

**Example**: Improving loop-based code

````r
# Slow approach: Growing vectors in a loop
slow_function <- function(n) {
  result <- c()
  for (i in 1:n) {
    result <- c(result, i^2)  # Growing vector - SLOW!
  }
  return(result)
}

# Time it
system.time(slow_function(10000))
````

````r
# AI-optimized version 1: Pre-allocate
fast_function_v1 <- function(n) {
  result <- numeric(n)  # Pre-allocate
  for (i in 1:n) {
    result[i] <- i^2
  }
  return(result)
}

# Time it
system.time(fast_function_v1(10000))
````

````r
# AI-optimized version 2: Vectorized (best!)
fast_function_v2 <- function(n) {
  result <- (1:n)^2
  return(result)
}

# Time it
system.time(fast_function_v2(10000))
````

## 5. Best Practices for AI-Assisted Coding

### ✅ Do's:

1. **Verify AI-generated code**: Always test and understand code before using it
2. **Provide context**: Give AI information about your data, goal, and constraints
3. **Iterate**: Refine prompts if the first result isn't perfect
4. **Learn from AI**: Use it as a teaching tool, not just a code generator
5. **Cite when appropriate**: If publishing, acknowledge AI assistance
6. **Use for boilerplate**: Let AI handle repetitive code patterns
7. **Ask for explanations**: Request comments and documentation

### ❌ Don'ts:

1. **Don't blindly trust**: AI can make mistakes or suggest poor practices
2. **Don't share sensitive data**: Be careful with proprietary or private information
3. **Don't rely completely**: Understand the fundamentals yourself
4. **Don't skip testing**: Always test AI-generated code thoroughly
5. **Don't ignore warnings**: If something seems wrong, it probably is
6. **Don't forget efficiency**: AI might not always give the most efficient solution
7. **Don't substitute learning**: Use AI to enhance, not replace, learning

## 7. AI Tools Comparison

### GitHub Copilot
- **Best for**: Real-time code completion in your IDE
- **Pros**: Fast, context-aware, integrates with VS Code/RStudio
- **Cons**: Subscription required, less explanatory

### ChatGPT / Claude
- **Best for**: Detailed explanations, complex problem-solving
- **Pros**: Great for learning, debugging, and understanding concepts
- **Cons**: Need to copy-paste code, no direct IDE integration

### When to use which:
- **Writing new code**: Copilot for boilerplate, ChatGPT for complex logic
- **Debugging**: ChatGPT/Claude for detailed analysis
- **Learning**: ChatGPT/Claude for explanations
- **Code completion**: Copilot while actively coding
