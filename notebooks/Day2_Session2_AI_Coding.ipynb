## 1. AI & Coding

If we go on any social media platform, we will see a lot of posts about "vibe coding". Although half of the posts are most likely exaggerated or misleading, it is undeniable that AI tools have democratized coding to a great extent.

AI tools can help with:
- **Code generation**: Writing code from natural language descriptions
- **Code completion & validation**: Suggesting code as you type, as well as finding and fixing errors
- **Code explanation**: Commenting the code
- **Documentation**: Writing comments and documentation

However, it is important to remember that AI tools <u> do make mistakes </u> and should not be blindly trusted. So, it is crucial to develop a "Trust but verify" mentality when using AI for coding.

## 2. Writing Effective Prompts for Code Generation

To efficiently use AI tools (for virtually any task), this first step is to write effective prompts. Good prompts should be:
- **Specific**: Clearly state what you want
- **Contextual**: Provide relevant information
- **Detailed**: Include requirements and constraints
- **Example-driven**: Show input/output examples when possible

**Example of Bad Prompt:**

"Write a function to get random numbers."

```python
import random

def get_random_number(min_value=0, max_value=100):
    return random.randint(min_value, max_value)

```

**Better Prompt:**

"Write an R function, called "simulate_random_numbers", which takes a minimum and a maximum and returns the random numbers"
```r
simulate_random_numbers <- function(min_value, max_value) {
  runif(1, min = min_value, max = max_value)
}

```

**Way Better Prompt:**
"Write an R function, called "simulate_random_numbers", which simulates numbers from a normal distribution, so it takes a mean and standard deviation values. Plus, it should also take the argument of N, which is the number of random numbers I want. The result should be a vector with the numbers only."

```r
simulate_random_numbers <- function(mean, sd, N) {
  rnorm(n = N, mean = mean, sd = sd)
}
```

One good way to improve your prompts is to treat them as a cowoker. Provide some context, explain the task - you do not need to give an extended abstract but, some information can be relevant for:
- Code creation
- Code review
- Providing relevant suggestions

**Example:**
"I am comparing cell counts between two groups, and I would like to simulate these counts with group information, to use it to optimize my code for visualization, using the package ggplot2. Provide code to perform this simulation."

*Function:*
```r
simulate_cell_counts <- function(
  N_per_group = 100,
  groups = c("Control", "Treatment"),
  means = c(50, 70),
  dispersion = 10
) {
  stopifnot(length(groups) == length(means))
  
  data <- lapply(seq_along(groups), function(i) {
    data.frame(
      group = groups[i],
      cell_count = rnbinom(
        n = N_per_group,
        mu = means[i],
        size = dispersion
      )
    )
  })
  
  do.call(rbind, data)
}
```
*Example usage:*
```r
set.seed(123)

sim_data <- simulate_cell_counts(
  N_per_group = 200,
  groups = c("Group A", "Group B"),
  means = c(40, 65),
  dispersion = 15
)

```
*Code for visualization:*
```r
library(ggplot2)

ggplot(sim_data, aes(x = group, y = cell_count, fill = group)) +
  geom_violin(trim = FALSE, alpha = 0.6) +
  geom_boxplot(width = 0.2, outlier.shape = NA) +
  theme_minimal() +
  labs(
    title = "Simulated Cell Counts by Group",
    y = "Cell count",
    x = "Group"
  )

```

**Why this structure works well for optimization**

- Distribution choice reflects real count data

You can also use AI tools to help you comment your code:

```r
# Complex code that might be confusing
library(dplyr)

result <- iris %>%
  group_by(Species) %>%
  summarise(
    n = n(),
    mean_petal_length = mean(Petal.Length),
    sd_petal_length = sd(Petal.Length),
    .groups = "drop"
  ) %>%
  mutate(
    cv = sd_petal_length / mean_petal_length * 100
  ) %>%
  arrange(desc(mean_petal_length))

print(result)
```

**AI Explanation of the above code:**

This code performs a grouped analysis of the iris dataset:

1. `group_by(Species)`: Groups data by iris species
2. `summarise()`: Calculates for each species:
   - `n()`: Count of observations
   - `mean(Petal.Length)`: Average petal length
   - `sd(Petal.Length)`: Standard deviation of petal length
3. `mutate()`: Adds coefficient of variation (CV) as percentage
4. `arrange(desc())`: Sorts by mean petal length in descending order

**Output**: A summary table showing species statistics ordered by petal length.

It can output the well-documented code:

```r
# Load the dplyr package for data manipulation
library(dplyr)

# Take the built-in iris dataset and perform a series of transformations
result <- iris %>%
  
  # Group the data by the Species column
  group_by(Species) %>%
  
  # For each species, compute summary statistics
  summarise(
    n = n(),                                      # Number of observations per species
    mean_petal_length = mean(Petal.Length),       # Mean petal length per species
    sd_petal_length = sd(Petal.Length),           # Standard deviation of petal length per species
    .groups = "drop"                              # Remove grouping after summarising
  ) %>%
  
  # Add a new column: coefficient of variation (CV) in percentage
  mutate(
    cv = sd_petal_length / mean_petal_length * 100
  ) %>%
  
  # Sort the results by mean petal length in descending order
  arrange(desc(mean_petal_length))

# Display the final summarized table
print(result)
```


## 3. Best Practices for AI-Assisted Coding

### ✅ Do's:

- **Provide context**: Give AI some information about your data, goal, and constraints. Do not overload it but, be specific
- **Use clear language**: Avoid ambiguity in your prompts
- **Iteration**: "If at first you don't succeed, try, try again"
- **Trust but verify**: Try your best to review to given code
- **Learn from AI**: Requesting explanations & documentation will also be the first step for you to learn new concepts
