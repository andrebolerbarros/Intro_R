## 1. Conditional Statements

Conditional statements allow your code to make decisions based on logical conditions.

### 1.1 If-Else Statement

```r
# Simple if statement
x <- 10

# if statement to check if x is greater than 5
if (x > 5) {
  print("x is greater than 5")
}
```

If you do not state what to do if the condition is false, `R` will not perform anything when the condition is false. However, if you want to specify an action when condition is false, you can use `else`.

```r
# if-else statement
temperature <- 25

if (temperature > 30) {
  print("It's hot!")
} else {
  print("It's comfortable")
}
```

You are not limited to a binary choice. You can use multiple `else if` statements to check for several conditions:

```r
# Multiple conditions
score <- 75

if (score >= 90) {
  grade <- "A"
} else if (score >= 80) {
  grade <- "B"
} else if (score >= 70) {
  grade <- "C"
} else if (score >= 60) {
  grade <- "D"
} else {
  grade <- "F"
}

print(paste("Grade:", grade))
```

For simple statements, `R` provides a shorthand notation using the `ifelse()` function:

```r
# Using ifelse for vectorized operations
scores <- c(95, 85, 72, 65, 55)
grades <- ifelse(scores >= 70, "Pass", "Fail")
print(data.frame(Score = scores, Grade = grades))
```

## 2. Loops

The goal of loops is to allow repeating code multiple times.

### 2.1 `for` Loops

For loops iterate over a sequence of values.

```r
# Simple for loop
for (i in 1:5) {
  print(i)
}
```

```r
# Loop through a vector
fruits <- c("apple", "banana", "cherry")
for (fruit in fruits) {
  print(paste("I like", fruit))
}
```

### 2.2 `while` Loops

While loops continue as long as a condition is TRUE.

```r
# While loop example
counter <- 1
while (counter <= 5) {
  print(paste("Counter:", counter))
  counter <- counter + 1
}
```

### 2.3 `repeat` Loops

Repeat loops continue indefinitely until explicitly broken.

```r
# Repeat loop with break
i <- 1
repeat {
  print(i)
  i <- i + 1
  if (i > 5) {
    break
  }
}
```

Alternatively, you can use the `break` statement to exit a loop when a certain condition is met.

```r
# Using break to exit a loop early
for (i in 1:10) {
  if (i == 6) {
    break
  }
  print(i)
}
```

You can also use the `next` statement to skip the current iteration and move to the next one.

```r
# Using next to skip an iteration
for (i in 1:10) {
  if (i %% 2 == 0) {  # Skip even numbers
    next
  }
  print(i)
}
```

### 2.4 `apply` Family Functions

The apply family (`apply`, `lapply`, `sapply`, `tapply`) are good alternatives to loops for applying functions over data structures.

```r
# Using sapply instead of a loop
numbers <- list(1:5, 6:10, 11:15)

# Calculate mean of each list element
means <- sapply(numbers, mean)
print(means)
```

## 3. Practical Examples

### 3.1 Finding Prime Numbers

### 3.2 Processing Data with Conditions

```r
# Check if a number is prime
is_prime <- function(n) {
  if (n <= 1) return(FALSE)
  if (n == 2) return(TRUE)
  
  for (i in 2:sqrt(n)) {
    if (n %% i == 0) {
      return(FALSE)
    }
  }
  return(TRUE)
}

# Find prime numbers between 1 and 20
primes <- c()
for (num in 1:20) {
  if (is_prime(num)) {
    primes <- c(primes, num)
  }
}
print(primes)
```

**Exercise 1**

Create a category based on a set of cut-offs (e.g. 50 & 75) 

```r
# Categorize data based on conditions
data <- data.frame(
  id = 1:10,
  value = c(12, 45, 78, 23, 89, 34, 67, 90, 15, 56)
)

# Add category based on value
data$category <- ifelse(data$value >= 75, "High",
                        ifelse(data$value >= 50, "Medium", "Low"))
print(data)
```
