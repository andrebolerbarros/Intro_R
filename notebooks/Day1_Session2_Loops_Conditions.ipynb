## 1. Conditional Statements

Conditional statements allow your code to make decisions based on logical conditions.

### 1.1 Basic if Statement

```python
# Simple if statement
x <- 10

if (x > 5) {
  print("x is greater than 5")
}
```

### 1.2 if-else Statement

```python
# if-else statement
temperature <- 25

if (temperature > 30) {
  print("It's hot!")
} else {
  print("It's comfortable")
}
```

### 1.3 if-else if-else Chain

```python
# Multiple conditions
score <- 75

if (score >= 90) {
  grade <- "A"
} else if (score >= 80) {
  grade <- "B"
} else if (score >= 70) {
  grade <- "C"
} else if (score >= 60) {
  grade <- "D"
} else {
  grade <- "F"
}

print(paste("Grade:", grade))
```

### 1.4 ifelse() Function for Vectors

The `ifelse()` function is vectorized and more efficient for working with vectors.

```python
# Using ifelse for vectorized operations
scores <- c(95, 85, 72, 65, 55)
grades <- ifelse(scores >= 70, "Pass", "Fail")
print(data.frame(Score = scores, Grade = grades))
```

## 2. Loops

Loops allow you to repeat code multiple times.

### 2.1 For Loops

For loops iterate over a sequence of values.

```python
# Simple for loop
for (i in 1:5) {
  print(i)
}
```

```python
# Loop through a vector
fruits <- c("apple", "banana", "cherry")
for (fruit in fruits) {
  print(paste("I like", fruit))
}
```

### 2.2 While Loops

While loops continue as long as a condition is TRUE.

```python
# While loop example
counter <- 1
while (counter <= 5) {
  print(paste("Counter:", counter))
  counter <- counter + 1
}
```

### 2.3 Repeat Loops

Repeat loops continue indefinitely until explicitly broken.

```python
# Repeat loop with break
i <- 1
repeat {
  print(i)
  i <- i + 1
  if (i > 5) {
    break
  }
}
```

### 2.4 Loop Control: break and next

```python
# Using break to exit a loop early
for (i in 1:10) {
  if (i == 6) {
    break
  }
  print(i)
}
```

```python
# Using next to skip an iteration
for (i in 1:10) {
  if (i %% 2 == 0) {  # Skip even numbers
    next
  }
  print(i)
}
```

## 3. Nested Loops

Loops can be nested inside other loops.

```python
# Nested for loops - multiplication table
for (i in 1:3) {
  for (j in 1:3) {
    result <- i * j
    print(paste(i, "x", j, "=", result))
  }
}
```

## 4. Practical Examples

### 4.1 Finding Prime Numbers

```python
# Check if a number is prime
is_prime <- function(n) {
  if (n <= 1) return(FALSE)
  if (n == 2) return(TRUE)
  
  for (i in 2:sqrt(n)) {
    if (n %% i == 0) {
      return(FALSE)
    }
  }
  return(TRUE)
}

# Find prime numbers between 1 and 20
primes <- c()
for (num in 1:20) {
  if (is_prime(num)) {
    primes <- c(primes, num)
  }
}
print(primes)
```

### 4.2 Processing Data with Conditions

```python
# Categorize data based on conditions
data <- data.frame(
  id = 1:10,
  value = c(12, 45, 78, 23, 89, 34, 67, 90, 15, 56)
)

# Add category based on value
data$category <- ifelse(data$value >= 75, "High",
                        ifelse(data$value >= 50, "Medium", "Low"))
print(data)
```

## 5. Vectorization: The R Way

R is designed for vectorized operations, which are often faster and cleaner than loops.

### 5.1 Vectorized vs Loop Approach

```python
# Loop approach (slower)
x <- 1:1000
result_loop <- numeric(length(x))
for (i in 1:length(x)) {
  result_loop[i] <- x[i]^2
}

# Vectorized approach (faster)
result_vec <- x^2

# Compare first 10 values
print(head(result_loop, 10))
print(head(result_vec, 10))
```

### 5.2 Apply Family Functions

The apply family (apply, lapply, sapply, tapply) provides alternatives to loops.

```python
# Using sapply instead of a loop
numbers <- list(1:5, 6:10, 11:15)

# Calculate mean of each list element
means <- sapply(numbers, mean)
print(means)
```

## 6. Common Pitfalls and Best Practices

### Tips:
1. **Use vectorized operations when possible** - They're faster and cleaner
2. **Pre-allocate memory** - If you must use loops, pre-allocate result vectors
3. **Avoid growing vectors in loops** - Use `result[i] <- value` instead of `result <- c(result, value)`
4. **Consider alternatives** - lapply, sapply, vapply are often better than for loops
5. **Use meaningful variable names** - `for (patient in patients)` is clearer than `for (i in 1:n)`
