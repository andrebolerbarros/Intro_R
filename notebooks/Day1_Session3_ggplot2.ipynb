## 1. Introduction to ggplot2

**ggplot2** is based on the "Grammar of Graphics" - a systematic way to build visualizations layer by layer.

### Basic Structure:
```r
ggplot(data, aes(x = x_var, y = y_var)) +
  geom_*() +
  theme_*() +
  labs()
```

**Key components:**
- **data**: Your dataset
- **aes()**: Aesthetic mappings (x, y, color, size, etc.)
- **geom_*()**: Geometric objects (points, lines, bars, etc.)
- **theme_*()**: Visual styling
- **labs()**: Labels and titles

````python
# Load necessary libraries
library(ggplot2)

# Load datasets
data(iris)
data(mpg)
````

## 2. Scatter Plots

### 2.1 Basic Scatter Plot

````python
# Basic scatter plot
ggplot(iris, aes(x = Sepal.Length, y = Sepal.Width)) +
  geom_point()
````

````python
# Add color by species
ggplot(iris, aes(x = Sepal.Length, y = Sepal.Width, color = Species)) +
  geom_point()
````

````python
# Customize point size and transparency
ggplot(iris, aes(x = Sepal.Length, y = Sepal.Width, color = Species)) +
  geom_point(size = 3, alpha = 0.6)
````

````python
# Different shapes by species
ggplot(iris, aes(x = Sepal.Length, y = Sepal.Width, color = Species, shape = Species)) +
  geom_point(size = 3)
````

### 2.2 Adding Trend Lines

````python
# Add linear regression line
ggplot(iris, aes(x = Sepal.Length, y = Sepal.Width)) +
  geom_point() +
  geom_smooth(method = "lm", se = TRUE)  # se = confidence interval
````

````python
# Separate lines by species
ggplot(iris, aes(x = Sepal.Length, y = Sepal.Width, color = Species)) +
  geom_point() +
  geom_smooth(method = "lm", se = FALSE)
````

````python
# Smooth curve (loess)
ggplot(iris, aes(x = Petal.Length, y = Petal.Width)) +
  geom_point(aes(color = Species)) +
  geom_smooth(method = "loess")
````

## 3. Bar Plots

### 3.1 Count Bar Plots

````python
# Simple bar plot (count)
ggplot(iris, aes(x = Species)) +
  geom_bar()
````

````python
# Add color
ggplot(iris, aes(x = Species, fill = Species)) +
  geom_bar()
````

### 3.2 Bar Plots with Values

````python
# Calculate mean values (we'll learn better ways tomorrow!)
setosa_mean <- mean(iris[iris$Species == "setosa", "Petal.Length"])
versicolor_mean <- mean(iris[iris$Species == "versicolor", "Petal.Length"])
virginica_mean <- mean(iris[iris$Species == "virginica", "Petal.Length"])

iris_summary <- data.frame(
  Species = c("setosa", "versicolor", "virginica"),
  mean_petal_length = c(setosa_mean, versicolor_mean, virginica_mean)
)

# Bar plot with values
ggplot(iris_summary, aes(x = Species, y = mean_petal_length, fill = Species)) +
  geom_col()  # geom_col() for pre-calculated values
````

## 4. Boxplots and Violin Plots

Great for comparing distributions across groups.

````python
# Basic boxplot
ggplot(iris, aes(x = Species, y = Petal.Length)) +
  geom_boxplot()
````

````python
# Add color and points
ggplot(iris, aes(x = Species, y = Petal.Length, fill = Species)) +
  geom_boxplot(alpha = 0.7) +
  geom_jitter(width = 0.2, alpha = 0.3)  # Jittered points
````

````python
# Violin plot (shows distribution shape)
ggplot(iris, aes(x = Species, y = Petal.Length, fill = Species)) +
  geom_violin()
````

````python
# Combine violin and boxplot
ggplot(iris, aes(x = Species, y = Petal.Length, fill = Species)) +
  geom_violin(alpha = 0.5) +
  geom_boxplot(width = 0.2, alpha = 0.7)
````

## 5. Histograms and Density Plots

For visualizing distributions.

````python
# Basic histogram
ggplot(iris, aes(x = Petal.Length)) +
  geom_histogram(bins = 30)
````

````python
# Histogram by species (stacked)
ggplot(iris, aes(x = Petal.Length, fill = Species)) +
  geom_histogram(bins = 30, alpha = 0.7)
````

````python
# Density plot
ggplot(iris, aes(x = Petal.Length, fill = Species)) +
  geom_density(alpha = 0.5)
````

````python
# Density plot with individual lines
ggplot(iris, aes(x = Petal.Length, color = Species)) +
  geom_density(size = 1.2)
````

## 6. Line Plots

Useful for time series or ordered data.

````python
# Create time series data
time_data <- data.frame(
  day = 1:30,
  temperature = 20 + 5 * sin((1:30) / 5) + rnorm(30, 0, 1)
)

# Basic line plot
ggplot(time_data, aes(x = day, y = temperature)) +
  geom_line()
````

````python
# Line and points
ggplot(time_data, aes(x = day, y = temperature)) +
  geom_line(color = "blue", size = 1) +
  geom_point(color = "red", size = 2)
````

## 7. Customizing Plots

### 7.1 Labels and Titles

````python
ggplot(iris, aes(x = Sepal.Length, y = Sepal.Width, color = Species)) +
  geom_point(size = 3) +
  labs(
    title = "Sepal Dimensions in Iris Species",
    subtitle = "Anderson's iris dataset",
    x = "Sepal Length (cm)",
    y = "Sepal Width (cm)",
    color = "Iris Species",
    caption = "Source: Fisher (1936)"
  )
````

### 7.2 Themes

````python
# Default theme
p <- ggplot(iris, aes(x = Species, y = Petal.Length, fill = Species)) +
  geom_boxplot()

p  # Default gray theme
````

````python
# Black and white theme
p + theme_bw()
````

````python
# Minimal theme (clean)
p + theme_minimal()
````

````python
# Classic theme
p + theme_classic()
````

### 7.3 Color Scales

````python
# Manual colors
ggplot(iris, aes(x = Sepal.Length, y = Sepal.Width, color = Species)) +
  geom_point(size = 3) +
  scale_color_manual(values = c("red", "blue", "green"))
````

````python
# Brewer color palettes (colorblind-friendly)
ggplot(iris, aes(x = Sepal.Length, y = Sepal.Width, color = Species)) +
  geom_point(size = 3) +
  scale_color_brewer(palette = "Set1")
````

````python
# Viridis color scale (perceptually uniform, colorblind-friendly)
ggplot(iris, aes(x = Sepal.Length, y = Sepal.Width, color = Petal.Length)) +
  geom_point(size = 3) +
  scale_color_viridis_c()
````

### 7.4 Axis Customization

````python
ggplot(iris, aes(x = Species, y = Petal.Length, fill = Species)) +
  geom_boxplot() +
  scale_y_continuous(
    limits = c(0, 8),
    breaks = seq(0, 8, 1)
  ) +
  theme_minimal()
````

## 8. Faceting (Multi-panel Plots)

Create separate plots for different subgroups.

````python
# Facet by one variable
ggplot(iris, aes(x = Sepal.Length, y = Sepal.Width)) +
  geom_point(aes(color = Species)) +
  geom_smooth(method = "lm", se = FALSE) +
  facet_wrap(~ Species)
````

````python
# Using mpg dataset - facet by two variables
ggplot(mpg, aes(x = displ, y = hwy)) +
  geom_point() +
  facet_grid(drv ~ cyl) +
  theme_bw()
````

````python
# Facet with free scales
ggplot(iris, aes(x = Sepal.Length)) +
  geom_histogram(bins = 20, fill = "steelblue", alpha = 0.7) +
  facet_wrap(~ Species, scales = "free_y") +
  theme_minimal()
````

## 9. Saving Plots

````python
# Create a plot
my_plot <- ggplot(iris, aes(x = Sepal.Length, y = Sepal.Width, color = Species)) +
  geom_point(size = 3) +
  theme_minimal() +
  labs(
    title = "Iris Sepal Dimensions",
    x = "Sepal Length (cm)",
    y = "Sepal Width (cm)"
  )

# Save as PNG
ggsave("iris_plot.png", plot = my_plot, width = 8, height = 6, dpi = 300)

# Save as PDF (vector format - best for publications)
ggsave("iris_plot.pdf", plot = my_plot, width = 8, height = 6)

print("Plots saved!")
````
