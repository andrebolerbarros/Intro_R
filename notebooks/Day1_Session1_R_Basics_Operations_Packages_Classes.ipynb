## 2. R Classes and Data Types

R has several basic data types:
- **numeric**: Numbers (integers and decimals)
  - **integer**: Whole numbers (no decimal points)
- **character**: Text strings
- **factor**: Categorical data (can be numbers or characters)
- **logical**: TRUE or FALSE

```r
# Numeric
height <- 175.5
class(height)

# Character
name <- "Alice"
class(name)

# Logical
is_student <- TRUE
class(is_student)

# Integer (use L suffix)
count <- 42L
class(count)
```

Another way to assess the class of an R object is to use the `typeof()` function, which provides more detailed information about the internal storage mode of the object.

```r
# Check the type of any object
typeof(height)
typeof(name)
typeof(is_student)
```

It is also possible to convert between data types via functions like `as.numeric()`, `as.character()`, `as.factor()`, and `as.logical()`. 

Be sure that change makes sense (e.g., converting characters to numeric will result in `NA` if the characters do not represent valid numbers) and that you understand how R handles such conversions (e.g. factors to numeric, by default, will give you the number corresponding to the observed level in factor).

```r
# Converting between types
x <- "123"
print(class(x))

x_numeric <- as.numeric(x)
print(class(x_numeric))

# Converting numeric to character
y <- 456
y_char <- as.character(y)
print(class(y_char))

# Converting factor directly to numeric
x = factor(c("red", "blue", "green", "blue"))
as.numeric(x)
```

## 3. Data Structures

### 3.1 Vectors

Vectors are the most basic data structure in R. They contain elements of the same type.

```r
# Creating vectors with c() function
ages <- c(25, 32, 28, 45, 38)
names <- c("Alice", "Bob", "Charlie", "Diana", "Eve")
is_phd <- c(TRUE, TRUE, FALSE, TRUE, FALSE)

print(ages)
print(names)
```

If you want to create a vector containing a sequence of numbers (e.g. from 1 to 10), instead of typing all the number, you can use the `:` operator or the `seq()` function:

```r
# Sequences
seq1 <- 1:10          # Simple sequence
seq2 <- seq(0, 100, by=10)  # Sequence with step
seq3 <- seq(0, 1, length.out=11)  # Sequence with length

print(seq1)
print(seq2)
print(seq3)
```

```r
?seq
```

Additionally, to repeat elements, you can use the `rep()` function:

```r
# Repetition
rep1 <- rep(5, times=10)  # Repeat 5 ten times
rep2 <- rep(c(1,2,3), times=3)  # Repeat vector three times
rep3 <- rep(c(1,2,3), each=3)   # Repeat each element three times

print(rep1)
print(rep2)
print(rep3)
```

#### 3.1.1 Vector Operations

R allows element-wise operations on vectors of the same length. 

**Note:** If vectors are of different lengths, R recycles the shorter vector to match the length of the longer one.

```r
# Arithmetic operations on vectors
v1 <- c(1, 2, 3, 4, 5)
v2 <- c(10, 20, 30, 40, 50)

v1 + v2  # Element-wise addition
v1 * v2  # Element-wise multiplication
v2 / v1  # Element-wise division

# !! Beware of this scenario !! #
y1 <- c(1,2,3)
y2 <- c(1,2,3,4,5,6)

y1 * y2  # R recycles y1 to match the length of y2
```

```r
# Operations with scalars (recycling)
v1 * 10  # Multiply each element by 10
v1 + 5   # Add 5 to each element
```

There are several useful functions that allow you to extract relevant information from vectors:

```r
# Useful vector functions
numbers <- c(12, 45, 23, 67, 34, 89, 21)

length(numbers)  # Number of elements
sum(numbers)     # Sum of all elements
mean(numbers)    # Average
median(numbers)  # Median
min(numbers)     # Minimum value
max(numbers)     # Maximum value
sd(numbers)      # Standard deviation
var(numbers)     # Variance

# Additional nugget - use the cat & paste functions for better output formatting
cat("The sum of numbers is:", sum(numbers), "\n")
```

#### 3.1.2 Indexing Vectors

You can access and modify specific elements in a vector using their indexes 

**Note:** R uses 1-based indexing

```r
fruits <- c("apple", "banana", "cherry", "date", "elderberry")

# Access by position
fruits[1]        # First element
fruits[3]        # Third element
fruits[c(1,3,5)] # Multiple elements
fruits[2:4]      # Range of elements

# Modify elements
fruits
fruits[2] <- "blueberry"
fruits
```

You can use the same logic to exclude elements by using a negative index:

```r
# Negative indexing (exclude elements)
fruits[-1]       # All except first
fruits[-c(2,4)]  # Exclude second and fourth
```

Finally, you can use logical vectors to get information from other vectors:

```r
# Logical indexing
ages <- c(25, 32, 28, 45, 38)
ages > 30        # Returns logical vector
ages[ages > 30]  # Filter values greater than 

```

#### 3.1.3 Named Vectors
You can assign names to the elements of a vector using the `names()` function. This is helpful for easier indexing, particularly when you do not know the exact position of an element.

```r
# Vectors can have names
person_ages <- c(Alice=25, Bob=32, Charlie=28)
print(person_ages)

# Access by name
person_ages["Alice"]

# Get names
names(person_ages)
```

### 3.2 Matrices

Matrices are 2-dimensional arrays with elements of the same type.

```r
# Creating matrices
mat1 <- matrix(1:12, nrow=3, ncol=4)
print(mat1)

# Fill by rows instead of columns
mat2 <- matrix(1:12, nrow=3, ncol=4, byrow=TRUE)
print(mat2)
```

As in vectors, you can access and modify specific elements in a matrix using their row and column indexes:

```r
# Matrix indexing: [row, column]
mat1[2, 3]      # Element in row 2, column 3
mat1[1, ]       # Entire first row
mat1[, 2]       # Entire second column
mat1[1:2, 2:3]  # Submatrix
```

`R` allows to collect information about matrices, the same way it does for vectors:

```r
# Matrix info
dim(mat1)    # Dimensions
nrow(mat1)   # Number of rows
ncol(mat1)   # Number of columns
summary(mat1)  # Summary statistics
```

### 3.2 Data Frames

Data frames are the most important data structure for data analysis - they're like spreadsheets!

```r
# Creating a data frame
df <- data.frame(
  name = c("Alice", "Bob", "Charlie", "Diana"),
  age = c(25, 32, 28, 45),
  department = c("Biology", "Chemistry", "Physics", "Biology"),
  is_phd = c(TRUE, TRUE, FALSE, TRUE)
)

df
```

Accessing and modifying elements in a data frame is similar to matrices.

```r
# Accessing rows & columns
df[1, ]         # First row
df[c(1,3), ]    # First and third rows
df[, 2]           # By column number
```

Additionally, you can use the `$` operator to access columns by name.

**Note:** beware of spaces & special characters in column names when using the `$` operator. You should always aim at having column names:
- starting with a letter
- no spaces (use `_` or `.` instead)
- no special characters (e.g. `-`, `/`, `\`, etc.)

```r
# Accessing columns
df$name           # Using $
df[["age"]]       # Using [[]]
df[, "department"] # Using [,]
```

As with vectors, you can use several functions to get information about data frames:

```r
# Filtering rows
df[df$age > 30, ]              # Rows where age > 30
df[df$department == "Biology", ]  # Rows from Biology department
```

Finally, several other function used for matrices can be used with data frames, such as `nrow()`, `ncol()`, `dim()`, `head()`, and `tail()`.

```r
nrow(df) #Get number of rows
ncol(df) #Get number of columns
dim(df)  #Get dimensions
```

```r
head(df) #First few rows
tail(df) #Last few rows
```

```r
str(df)  # Structure of the data frame
summary(df)  # Summary statistics of the data frame
```

### 3.3 Lists

Lists can contain elements of different types, including single objects, vectors, matrices, data frames, and even other lists.

```r
# Creating a list
my_list <- list(
  name = "John Doe",
  age = 35,
  scores = c(85, 92, 78),
  is_phd = TRUE
)

print(my_list)
```

```r
# Accessing list elements
my_list$name        # By name using $
my_list[["age"]]    # By name using [[]]
my_list[[2]]        # By position
my_list[1]          # Returns a list with one element
```

## 5. R Packages and Libraries

**R packages** are collections of functions, data, and documentation that extend R's capabilities. Think of them as plugins you add to R:
- **Base R** (package that comes by default) provides the backbone functions
- **Packages** add specialized tools for specific tasks

### 5.1. Where to Find R Packages?

**[1. CRAN (Comprehensive R Archive Network)](https://cran.r-project.org/)**
- Official repository with over 18,000 packages
- All packages tested and documented

**[2. Bioconductor](https://bioconductor.org/)**
- Specialized for bioinformatics and genomics
- Over 2,000 packages for biological data analysis
- Website: 

**[3. GitHub](https://github.com)**
- Development versions and specialized packages
- <u>Not always as stable as CRAN packages</u>

### 5.2 Installing Packages

To install packages in the R environment, use `install.packages()` to install from CRAN. 

```r
# Install a single package
install.packages("ggplot2")

# Install multiple packages at once
install.packages(c("dplyr", "tidyr", "readr"))

# Install from a specific CRAN mirror
install.packages("package_name", repos = "https://cran.rstudio.com/")

# Note: During installation, you may be asked to choose a CRAN mirror
# Select one close to your location for faster downloads
```

### 5.3 Loading Packages

After installation, use `library()` or `require()` to load a package. **You must load packages in each R session!**

```r
# Load a package (preferred method)
library(ggplot2)
```

`R` also provides some relevant functions to manage packages:

```r
# See all installed packages
# installed.packages()

# See loaded packages in current session
# search()

# Get help for a package
help(package = "ggplot2")

# Update all packages
# update.packages()

# Update a specific package
# update.packages("ggplot2")

# Remove a package
# remove.packages("package_name")

# Check package version
packageVersion("ggplot2")
```
