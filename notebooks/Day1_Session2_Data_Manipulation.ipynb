### 1.2 Reading CSV Files

CSV (Comma-Separated Values) is the most common data format.

```python
# First, let's create a sample CSV file
sample_data <- data.frame(
  patient_id = 1:10,
  age = c(25, 34, 45, 23, 67, 54, 32, 29, 41, 38),
  treatment = rep(c("A", "B"), each=5),
  response = c(7.2, 8.1, 6.5, 7.8, 5.4, 8.9, 9.2, 7.1, 6.8, 8.5)
)

# Write to CSV
write.csv(sample_data, "sample_data.csv", row.names=FALSE)
print("CSV file created!")
```

```python
# Read CSV file
data <- read.csv("sample_data.csv")
head(data)
```

```python
# read.csv() options
# header=TRUE/FALSE : Does the file have column names?
# sep="," : What's the delimiter? (could be "\t" for tab, ";" for semicolon)
# na.strings="NA" : What represents missing values?
# stringsAsFactors=FALSE : Should character columns be converted to factors?

# Example:
data <- read.csv("sample_data.csv", header=TRUE, stringsAsFactors=FALSE)
str(data)
```

## 3. Subsetting Data Frames

### 3.1 Selecting Columns

```python
# Select single column (returns a vector)
iris$Sepal.Length
iris[["Sepal.Length"]]
iris[, "Sepal.Length"]
iris[, 1]
```

```python
# Select multiple columns (returns a data frame)
iris[, c("Sepal.Length", "Sepal.Width")]
iris[, c(1, 2)]
iris[, 1:3]
```

### 3.2 Selecting Rows

```python
# Select rows by index
iris[1:5, ]         # First 5 rows
iris[c(1,5,10), ]   # Specific rows
```

```python
# Filter rows based on conditions
# Setosa species only
setosa <- iris[iris$Species == "setosa", ]
head(setosa)
```

```python
# Multiple conditions
# Setosa with Sepal.Length > 5
large_setosa <- iris[iris$Species == "setosa" & iris$Sepal.Length > 5, ]
head(large_setosa)
```

```python
# OR condition: |
# Setosa or versicolor
two_species <- iris[iris$Species == "setosa" | iris$Species == "versicolor", ]
table(two_species$Species)
```

```python
# Using %in% for multiple values
selected <- iris[iris$Species %in% c("setosa", "versicolor"), ]
table(selected$Species)
```

### 3.3 subset() Function

A cleaner way to filter data.

```python
# Using subset()
setosa_sub <- subset(iris, Species == "setosa")
head(setosa_sub)

# With multiple conditions
large_setosa_sub <- subset(iris, Species == "setosa" & Sepal.Length > 5)
head(large_setosa_sub)
```

```python
# Select specific columns while filtering
setosa_sepal <- subset(iris, 
                       Species == "setosa", 
                       select = c(Sepal.Length, Sepal.Width))
head(setosa_sepal)
```

## 6. Handling Missing Data

### 6.1 Identifying Missing Data

```python
# Create data with missing values
data_na <- data.frame(
  id = 1:8,
  value1 = c(10, 20, NA, 40, 50, NA, 70, 80),
  value2 = c(5, NA, 15, 20, NA, 30, 35, 40),
  group = c("A", "B", "A", NA, "B", "A", "B", "A")
)

print(data_na)
```

```python
# Check for NA values
is.na(data_na)          # Returns logical matrix
sum(is.na(data_na))     # Total number of NAs
colSums(is.na(data_na)) # NAs per column
```

```python
# Check for complete cases (rows with no NAs)
complete.cases(data_na)
sum(complete.cases(data_na))  # Number of complete rows
```

### 6.2 Removing Missing Data

```python
# Remove rows with any NA
data_complete <- na.omit(data_na)
print(data_complete)

# Or using complete.cases()
data_complete2 <- data_na[complete.cases(data_na), ]
print(data_complete2)
```

```python
# Remove rows with NA in specific column
data_no_na_value1 <- data_na[!is.na(data_na$value1), ]
print(data_no_na_value1)
```

### 6.3 Replacing Missing Data

```python
# Replace NA with a specific value
data_replaced <- data_na
data_replaced$value1[is.na(data_replaced$value1)] <- 0
print(data_replaced)
```

```python
# Replace NA with mean (for numeric columns)
data_replaced2 <- data_na
mean_value2 <- mean(data_replaced2$value2, na.rm=TRUE)
data_replaced2$value2[is.na(data_replaced2$value2)] <- mean_value2
print(data_replaced2)
```

## 7. Aggregating Data

### 7.1 Basic Aggregation

```python
# Calculate mean by species
aggregate(Sepal.Length ~ Species, data=iris, FUN=mean)
```

```python
# Multiple columns
aggregate(cbind(Sepal.Length, Sepal.Width) ~ Species, data=iris, FUN=mean)
```

```python
# Different functions
aggregate(Sepal.Length ~ Species, data=iris, FUN=sd)      # Standard deviation
aggregate(Sepal.Length ~ Species, data=iris, FUN=median)  # Median
aggregate(Sepal.Length ~ Species, data=iris, FUN=length)  # Count
```

### 7.2 Table Functions

```python
# Frequency table
table(iris$Species)
```

```python
# Cross-tabulation
# Create a size category
iris$Size <- ifelse(iris$Petal.Length > median(iris$Petal.Length), "Large", "Small")
table(iris$Species, iris$Size)
```

```python
# Proportions
prop.table(table(iris$Species))  # Proportions
prop.table(table(iris$Species)) * 100  # Percentages
```

## 8. Combining Data Frames

### 8.1 Binding Rows and Columns

```python
# Create two small data frames
df1 <- data.frame(id=1:3, value=c(10, 20, 30))
df2 <- data.frame(id=4:6, value=c(40, 50, 60))

# Row binding (stacking)
df_rows <- rbind(df1, df2)
print(df_rows)
```

```python
# Column binding (side by side)
df3 <- data.frame(name=c("A", "B", "C"))
df_cols <- cbind(df1, df3)
print(df_cols)
```

### 8.2 Merging Data Frames

```python
# Create two related data frames
patients <- data.frame(
  patient_id = 1:5,
  age = c(45, 32, 56, 28, 41)
)

measurements <- data.frame(
  patient_id = c(1, 2, 3, 3, 4),
  measurement = c(7.2, 8.1, 6.5, 7.8, 5.4)
)

print(patients)
print(measurements)
```

```python
# Inner join (keep only matching rows)
merged <- merge(patients, measurements, by="patient_id")
print(merged)
```

```python
# Left join (keep all from left)
merged_left <- merge(patients, measurements, by="patient_id", all.x=TRUE)
print(merged_left)
```
