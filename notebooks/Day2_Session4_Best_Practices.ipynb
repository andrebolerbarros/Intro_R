## 1. Reproducible Research

### 1.1 Why Reproducibility Matters

**Reproducible research** means that:
- Others (and your future self!) can recreate your results
- Your analysis is transparent and verifiable
- Errors can be easily detected and corrected
- Your work has greater impact and credibility

**Key principles:**
1. Document everything
2. Use version control (Git)
3. Keep data and analysis together
4. Make code readable
5. Share your work

## 2. Project Organization

### 2.1 R Projects

**Always use RStudio Projects (.Rproj)**

Benefits:
- Sets working directory automatically
- Keeps related files together
- Makes paths relative (portable)
- Easier collaboration

**Create a new project:**
File → New Project → New Directory → New Project

### 2.2 Folder Structure

A well-organized project:

```
my_project/
│
├── my_project.Rproj       # RStudio project file
├── README.md              # Project description
├── data/                  # Raw data (read-only!)
│   ├── raw/              
│   └── processed/        
├── scripts/               # R scripts
│   ├── 01_data_cleaning.R
│   ├── 02_analysis.R
│   └── 03_visualization.R
├── notebooks/             # Jupyter notebooks or RMarkdown
├── output/                # Generated files
│   ├── figures/
│   └── tables/
└── functions/             # Custom functions
```

### 3.3 Writing Functions

````r
# If you copy-paste code more than twice, write a function!

# Calculate standard error
standard_error <- function(x, na.rm = TRUE) {
  # Calculate standard error of the mean
  # Args:
  #   x: numeric vector
  #   na.rm: remove NA values? (default TRUE)
  # Returns:
  #   standard error (numeric)
  
  if (!is.numeric(x)) {
    stop("x must be numeric")
  }
  
  sd(x, na.rm = na.rm) / sqrt(length(x[!is.na(x)]))
}

# Test it
test_data <- c(10, 12, 15, 18, 20)
standard_error(test_data)
````

## 5. R Markdown Basics

R Markdown lets you combine code, results, and narrative text in one document.

### 5.1 Basic Structure

```markdown
---
title: "My Analysis"
author: "Your Name"
date: "2024-01-11"
output: html_document
---

## Introduction

This is regular text.

```{r}
# This is R code
summary(iris)
```

## Results

More text here.
```

**Keyboard shortcut to insert code chunk:** Ctrl+Alt+I (Windows) or Cmd+Option+I (Mac)

### 5.2 Chunk Options

```markdown
```{r, echo=FALSE}
# Code runs but doesn't show in output
```

```{r, eval=FALSE}
# Code shows but doesn't run
```

```{r, message=FALSE, warning=FALSE}
# Suppress messages and warnings
```

```{r, fig.width=8, fig.height=6}
# Control figure size
```
```

## 7. Version Control with Git

### 7.1 Why Use Git?

- Track changes over time
- Collaborate with others
- Revert to previous versions
- Backup your work

### 7.2 Basic Git Workflow

```bash
# Initialize repository
git init

# Check status
git status

# Stage files
git add script.R

# Commit changes
git commit -m "Add data cleaning script"

# Push to GitHub
git push origin main
```

**Use RStudio's Git integration!**
- Tools → Version Control → Project Setup

### 7.3 .gitignore

Tell Git which files to ignore:

```

# Temporary files

## 10. Checklist for Reproducible Analysis

Before sharing your analysis, check:

- [ ] Code runs from start to finish without errors
- [ ] All required packages are listed and loaded
- [ ] No absolute file paths (use relative paths or `here()`)
- [ ] No `setwd()` commands
- [ ] Random seeds set where needed
- [ ] Clear comments explaining key decisions
- [ ] README file describes project
- [ ] Data files documented (source, variables)
- [ ] Results and figures can be regenerated
- [ ] Session info included (`sessionInfo()`)
