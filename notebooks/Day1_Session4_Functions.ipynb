## 1. What Are Functions?

Functions are reusable blocks of code that perform specific tasks. They help you:
- **Avoid repetition** (DRY principle: Don't Repeat Yourself)
- **Organize code** into logical units
- **Make code more readable** and maintainable
- **Reduce errors** by centralizing logic
- **Share code** with others

### Function Anatomy

```r
function_name <- function(arguments) {
  # Function body
  # Code that does something
  return(result)  # Optional: explicitly return a value
}
```

## 3. Writing Your First Function

### 3.1 Simple Functions

````r
# Simple function with no arguments
greet <- function() {
  print("Hello, R programmer!")
}

# Call the function
greet()
````

````r
# Function with one argument
greet_person <- function(name) {
  message <- paste("Hello,", name, "!")
  print(message)
}

greet_person("Alice")
greet_person("Bob")
````

````r
# Function with multiple arguments
calculate_bmi <- function(weight_kg, height_m) {
  bmi <- weight_kg / (height_m^2)
  return(bmi)
}

# Test the function
calculate_bmi(70, 1.75)
calculate_bmi(80, 1.80)
````

### 3.2 Default Arguments

You can provide default values for arguments.

````r
# Function with default arguments
greet_person_default <- function(name = "Friend", time = "day") {
  message <- paste("Good", time, ",", name, "!")
  print(message)
}

# Call with no arguments (uses defaults)
greet_person_default()

# Call with one argument
greet_person_default("Alice")

# Call with both arguments
greet_person_default("Bob", "evening")

# Use named arguments (can be in any order)
greet_person_default(time = "morning", name = "Charlie")
````

### 3.3 Return Values

Functions can return values explicitly or implicitly.

````r
# Explicit return
square_explicit <- function(x) {
  result <- x^2
  return(result)
}

# Implicit return (last evaluated expression)
square_implicit <- function(x) {
  x^2  # This is automatically returned
}

# Both work the same way
square_explicit(5)
square_implicit(5)
````

````r
# Return multiple values using a list
calculate_stats <- function(x) {
  result <- list(
    mean = mean(x),
    median = median(x),
    sd = sd(x),
    min = min(x),
    max = max(x)
  )
  return(result)
}

# Test the function
data <- c(10, 20, 30, 40, 50)
stats <- calculate_stats(data)
print(stats)

# Access individual elements
stats$mean
stats$sd
````

## 4. Function Scope and Variables

Variables defined inside a function are **local** to that function.

````r
# Demonstrating scope
x <- 10  # Global variable

test_scope <- function() {
  x <- 20  # Local variable (doesn't affect global x)
  y <- 30  # Local variable
  print(paste("Inside function: x =", x, ", y =", y))
}

test_scope()
print(paste("Outside function: x =", x))
# print(y)  # This would cause an error - y doesn't exist outside the function
````

````r
# Accessing global variables (but avoid modifying them!)
global_value <- 100

use_global <- function() {
  result <- global_value * 2  # Can read global variables
  return(result)
}

use_global()
````

## 5. Practical Function Examples

### 5.1 Data Processing Function

````r
# Function to standardize (z-score) a numeric vector
standardize <- function(x) {
  # Remove NA values
  x_clean <- x[!is.na(x)]
  
  # Calculate z-scores
  z_scores <- (x_clean - mean(x_clean)) / sd(x_clean)
  
  return(z_scores)
}

# Test the function
data <- c(10, 20, 30, 40, 50, NA, 60)
standardized <- standardize(data)
print(standardized)

# Verify: mean should be ~0, sd should be ~1
mean(standardized)
sd(standardized)
````

### 5.2 Conditional Logic in Functions

````r
# Function to categorize BMI
categorize_bmi <- function(bmi) {
  if (bmi < 18.5) {
    category <- "Underweight"
  } else if (bmi < 25) {
    category <- "Normal weight"
  } else if (bmi < 30) {
    category <- "Overweight"
  } else {
    category <- "Obese"
  }
  return(category)
}

# Test the function
categorize_bmi(22)
categorize_bmi(28)
categorize_bmi(32)
````

````r
# Vectorized version using ifelse
categorize_bmi_vectorized <- function(bmi) {
  categories <- ifelse(bmi < 18.5, "Underweight",
                ifelse(bmi < 25, "Normal weight",
                ifelse(bmi < 30, "Overweight", "Obese")))
  return(categories)
}

# Test with multiple values
bmis <- c(17, 22, 28, 32, 35)
categorize_bmi_vectorized(bmis)
````

### 5.3 Functions with Validation

````r
# Function with input validation
safe_divide <- function(numerator, denominator) {
  # Check for division by zero
  if (denominator == 0) {
    stop("Error: Cannot divide by zero!")
  }
  
  # Check for non-numeric inputs
  if (!is.numeric(numerator) || !is.numeric(denominator)) {
    stop("Error: Both arguments must be numeric!")
  }
  
  result <- numerator / denominator
  return(result)
}

# Test the function
safe_divide(10, 2)
# safe_divide(10, 0)    # This will produce an error message
# safe_divide(10, "a")  # This will also produce an error
````

````r
# Function with warnings
calculate_mean_with_warning <- function(x) {
  # Check for NA values
  if (any(is.na(x))) {
    warning("Data contains NA values. These will be removed.")
    x <- x[!is.na(x)]
  }
  
  # Check for empty vector
  if (length(x) == 0) {
    stop("Error: Cannot calculate mean of empty vector!")
  }
  
  return(mean(x))
}

# Test with NA values
data_with_na <- c(1, 2, 3, NA, 5)
calculate_mean_with_warning(data_with_na)
````

## 6. Advanced Function Concepts

### 6.1 Anonymous Functions (Lambda Functions)

Functions without names, useful for one-time operations.

````r
# Using anonymous functions with lapply
numbers <- list(1:5, 6:10, 11:15)

# Calculate sum of each list element using anonymous function
sums <- lapply(numbers, function(x) sum(x))
print(sums)

# Calculate mean using anonymous function
means <- lapply(numbers, function(x) mean(x))
print(means)
````

### 6.2 Functions as Arguments

Functions can be passed as arguments to other functions.

````r
# Function that applies another function to data
apply_operation <- function(x, operation) {
  result <- operation(x)
  return(result)
}

# Test with different operations
data <- c(1, 2, 3, 4, 5)

apply_operation(data, mean)
apply_operation(data, median)
apply_operation(data, sd)

# With a custom function
double <- function(x) x * 2
apply_operation(data, double)
````

### 6.3 The ... Argument (Ellipsis)

Pass additional arguments to functions inside your function.

````r
# Function that passes additional arguments
flexible_summary <- function(x, ...) {
  # Calculate various statistics
  # ... allows passing additional arguments to internal functions
  m <- mean(x, ...)
  med <- median(x, ...)
  
  result <- list(
    mean = m,
    median = med
  )
  
  return(result)
}

# Test with data containing NA
data_na <- c(1, 2, 3, NA, 5)

# Without na.rm argument (will return NA)
flexible_summary(data_na)

# With na.rm argument (passed through ...)
flexible_summary(data_na, na.rm = TRUE)
````

## 7. Function Best Practices

### Key Principles:

1. **Use descriptive names**: `calculate_mean()` not `cm()`
2. **One function, one task**: Keep functions focused
3. **Document your functions**: Add comments explaining what they do
4. **Validate inputs**: Check for errors early
5. **Use meaningful variable names**: Inside and outside the function
6. **Test your functions**: With various inputs, including edge cases
7. **Keep functions short**: If it's too long, break it into smaller functions

### 7.1 Documenting Functions

Good documentation helps others (and future you) understand your code.

````r
#' Calculate Body Mass Index (BMI)
#'
#' This function calculates BMI from weight and height.
#'
#' @param weight_kg Numeric. Weight in kilograms
#' @param height_m Numeric. Height in meters
#'
#' @return Numeric. BMI value
#'
#' @examples
#' calculate_bmi_documented(70, 1.75)
#' calculate_bmi_documented(80, 1.80)
#'
calculate_bmi_documented <- function(weight_kg, height_m) {
  # Validate inputs
  if (!is.numeric(weight_kg) || !is.numeric(height_m)) {
    stop("Both weight and height must be numeric")
  }
  
  if (weight_kg <= 0 || height_m <= 0) {
    stop("Weight and height must be positive")
  }
  
  # Calculate BMI
  bmi <- weight_kg / (height_m^2)
  
  # Round to 2 decimal places
  bmi <- round(bmi, 2)
  
  return(bmi)
}

# Test the function
calculate_bmi_documented(70, 1.75)
````

## 8. Practical Project: Data Analysis Functions

Let's build a small library of data analysis functions.

````r
# Function 1: Remove outliers using IQR method
remove_outliers <- function(x, threshold = 1.5) {
  # Calculate quartiles
  q1 <- quantile(x, 0.25, na.rm = TRUE)
  q3 <- quantile(x, 0.75, na.rm = TRUE)
  iqr <- q3 - q1
  
  # Define bounds
  lower_bound <- q1 - threshold * iqr
  upper_bound <- q3 + threshold * iqr
  
  # Filter data
  clean_data <- x[x >= lower_bound & x <= upper_bound]
  
  # Report
  n_removed <- length(x) - length(clean_data)
  if (n_removed > 0) {
    message(paste("Removed", n_removed, "outliers"))
  }
  
  return(clean_data)
}

# Test
data_with_outliers <- c(1, 2, 3, 4, 5, 100)  # 100 is an outlier
clean_data <- remove_outliers(data_with_outliers)
print(clean_data)
````

````r
# Function 2: Comprehensive descriptive statistics
describe_data <- function(x) {
  # Remove NA values
  x_clean <- x[!is.na(x)]
  n_na <- sum(is.na(x))
  
  # Calculate statistics
  stats <- list(
    n = length(x_clean),
    n_missing = n_na,
    mean = mean(x_clean),
    median = median(x_clean),
    sd = sd(x_clean),
    min = min(x_clean),
    max = max(x_clean),
    q25 = quantile(x_clean, 0.25),
    q75 = quantile(x_clean, 0.75)
  )
  
  # Create a nice data frame
  result <- data.frame(
    Statistic = names(stats),
    Value = unlist(stats)
  )
  
  return(result)
}

# Test
sample_data <- c(10, 20, 30, 40, 50, NA, 60, 70)
describe_data(sample_data)
````

````r
# Function 3: Quick data quality check
check_data_quality <- function(df) {
  # Initialize results
  results <- data.frame(
    Column = names(df),
    Type = sapply(df, class),
    N_Missing = sapply(df, function(x) sum(is.na(x))),
    Pct_Missing = sapply(df, function(x) round(sum(is.na(x))/length(x)*100, 2)),
    N_Unique = sapply(df, function(x) length(unique(x)))
  )
  
  rownames(results) <- NULL
  return(results)
}

# Test with iris dataset
check_data_quality(iris)
````
