## 1. Project Organization

A well-organized project structure is essential for reproducible research. Allows you and others to easily navigate your data, code, results, even relevant documentation. Here is a common project structure:

```
ProjectID/
    ├── data/
    │   ├── 00_rawdata
    │   ├── 01_filtered/
    │   │   ├── 01a_filtered_relax
    │   │   └── 01b_filtered_strict
    │   ├── 02_aligned
    │   └── ...
    ├── docs/
    │   ├── 00_ExperimentalInfo
    │   ├── 01_SequencingReports
    │   ├── ...
    │   └── 99_RelevantReferences
    └── workflow/
        ├── 98_notebooks
        └── 99_scripts
```

Make sure that:
- Use clear and descriptive names for folders
- Do not repeat names (e.g., avoid "data/data")
- Select a consistent naming convention (e.g., raw_data vs RawData)
- If you want to add dates to file names, use the `YYYYMMDD` format for better sorting
- Add a `README` file to explain the project structure and contents

Documenting & managing your programming environment ensures that your code runs consistently across different systems and over time.

## 3. Coding & Documenting Using the Right Tools

To produce a truly reproducible code, you need to work on it in two fronts:

### 3.1 Coding Best Practices

Following these practices greatly improve the quality of your code:
- Create good, clear and consistent code style (with proper naming conventions, spacing, etc.);
- Write modular code by breaking down your code into smaller, reusable functions;
- Comment your code to explain complex logic or decisions;
- Avoid hardcoding values!
- Test your code to ensure it works properly

Here are some of the examples of these practices:

Another relevant practice is, instead of writing long scripts with "copy-paste" code, you should create functions to perform these repetitive tasks:

`````r
# Calculate standard error
standard_error <- function(x, na.rm = TRUE) {
  # Calculate standard error of the mean
  # Args:
  #   x: numeric vector
  #   na.rm: remove NA values? (default TRUE)
  # Returns:
  #   standard error (numeric)
  
  if (!is.numeric(x)) {
    stop("x must be numeric")
  }
  
  sd(x, na.rm = na.rm) / sqrt(length(x[!is.na(x)]))
}

# Test it
test_data <- c(10, 12, 15, 18, 20)
standard_error(test_data)
`````

You should comment your function code with some information about its purpose, inputs and outputs. <u>Ideally, you should create a proper function file</u> but you can simplify this task.

### 3.2 Literate Programming with R Markdown

Quarto notebooks or Jupyter notebooks are very good examples of literate programming - combining code, results, and narrative text in one document.

It starts with a simple header:

```markdown
---
title: "My Analysis"
author: "Your Name"
date: "2024-01-11"
output: html_document
---
````

Then, you can start writing your text and code chunks
As a suggestion, add a top chunk for loading all the libraries you will use:

**Show code, don't run:**

## 4. Use a Version Control System

Version control is essential for tracking changes, collaborating, and maintaining a history of your work. To goal is to allow you to have records of project changes, file modifications, etc., over time. 

You can do this in your computer manually (e.g. using dates in file names), but this can quickly become unmanageable. Instead, you can use a version control system like [Github](https://github.com/).

### 4.1 Github!

Github is a web-based platform for version control. It provides a user-friendly interface for managing your code respositories in a very easy way. Plus, it is well integrated with RStudio, VSCode, JupyterLab,... .

Its main features are:
- Track changes over time
- Collaborate with others
- Revert to previous versions
- Backup your work

Repository example: [This course!](https://github.com/andrebolerbarros/Intro_R)

To learn more about Github, check these links:
- https://docs.github.com/en/get-started/start-your-journey/hello-world
- https://rogerdudler.github.io/git-guide/

There are other version control systems available (e.g. [Gitlab](https://about.gitlab.com/)) but, if you know Github, you can easily adapt to others.


### 4.2. Scientific repositories

Once you have your code and data at a ready stage (does not mean final or perfect!), you can upload it to a scientific repository to share with the community. There are some popular options:
- **[Figshare](https://figshare.com/)**: For sharing datasets and research outputs
- **[Zenodo](https://zenodo.org/)**: Long-term archival for research data and code

These indeed allow for version control (although you should have the same number of versions here than on Github) and has the additional benefit of providing a DOI for your work, making it citable and, thus, increasing its visibility and impact.

However, **beware of sensitive data!** - Not everything is shareable (at least "as is")!.
